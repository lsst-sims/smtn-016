```python
import rubin_sim.maf as maf
import matplotlib.pylab as plt
%matplotlib inline
from rubin_sim.data import get_baseline
import os
```

```python
baseline_file = 'baseline_v5.3.5_10yrs.db'
runName = os.path.split(baseline_file)[-1].replace('.db', '')
out_dir='temp'
```

```python
bundleList = []
summary_list = [maf.metrics.MedianMetric()]
slicer = maf.slicers.HealpixSlicer()
for filtername in 'ugrizy':
    sql = 'band = "%s"' % filtername
    metric = maf.metrics.SurfaceBrightLimitMetric(metric_name='sb limit, %s' % filtername)
    bundleList.append(maf.MetricBundle(metric, slicer, constraint=sql, summary_metrics=summary_list, run_name=runName))
```

```python
bg = maf.MetricBundleGroup(bundleList, baseline_file, out_dir, None)
bg.run_all()
bg.plot_all(closefigs=False)
```

```python
for b in bundleList:
    print(b.constraint, b.summary_values)
```
